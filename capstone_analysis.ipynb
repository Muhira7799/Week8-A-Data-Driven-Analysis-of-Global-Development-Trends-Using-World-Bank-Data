```python
clean_data = analysis_data.copy()

analysis_years = [str(year) for year in range(2010, 2025)]

for indicator in clean_data["Indicator Code"].unique():
    
    mask = clean_data["Indicator Code"] == indicator
    
    clean_data.loc[mask, analysis_years] = (
        clean_data.loc[mask, analysis_years]
        .interpolate(axis=1, limit_area="inside")
    )

print("Missing values after interpolation:")

print(
    clean_data[analysis_years]
    .isna()
    .sum()
)
```

```python
long_data = clean_data.melt(
    id_vars=[
        "Country Name",
        "Country Code",
        "Indicator Name",
        "Indicator Code"
    ],
    value_vars=[str(year) for year in range(2010, 2025)],
    var_name="Year",
    value_name="Value"
)

long_data.head(10)
```

```python
long_data["Year"] = long_data["Year"].astype(int)

print("Rows:", len(long_data))
print("Columns:", long_data.columns.tolist())
```

```python
indicator_names = {
    "IT.NET.USER.ZS": "Internet Users (%)",
    "IT.CEL.SETS.P2": "Mobile Subscriptions",
    "IT.NET.BBND.P2": "Fixed Broadband"
}

long_data["Indicator"] = long_data["Indicator Code"].map(indicator_names)

long_data.head(10)
```

```python


print(long_data["Indicator"].value_counts())
```

```python
feature_data = long_data.pivot_table(
    index=["Country Name", "Country Code", "Year"],
    columns="Indicator",
    values="Value"
).reset_index()

feature_data["Internet-Mobile Gap"] = (
    feature_data["Mobile Subscriptions"]
    - feature_data["Internet Users (%)"]
)

print(feature_data.head(10))
```

```python
print(feature_data.columns)
```

```python
feature_data["Internet-Broadband Gap"] = (
    feature_data["Internet Users (%)"]
    - feature_data["Fixed Broadband"]
)

print(
    feature_data[
        [
            "Country Name",
            "Year",
            "Internet Users (%)",
            "Fixed Broadband",
            "Internet-Broadband Gap"
        ]
    ].head(10)
)
```

```python
score_columns = [
    "Internet Users (%)",
    "Mobile Subscriptions",
    "Fixed Broadband"
]

for col in score_columns:
    min_value = feature_data[col].min()
    max_value = feature_data[col].max()

    feature_data[f"{col} Normalized"] = (
        (feature_data[col] - min_value)
        / (max_value - min_value)
    )

feature_data["Digital Connectivity Score"] = (
    feature_data["Internet Users (%) Normalized"]
    + feature_data["Mobile Subscriptions Normalized"]
    + feature_data["Fixed Broadband Normalized"]
) / 3

print(
    feature_data[
        [
            "Country Name",
            "Year",
            "Internet Users (%)",
            "Mobile Subscriptions",
            "Fixed Broadband",
            "Digital Connectivity Score"
        ]
    ].head(10)
)
```

### Data Analysis

1- Trend and Patterns
The analysis examines changes in digital connectivity across 54 African countries from 2010 to 2024. The main indicators analyzed are Internet Users (%), Mobile Subscriptions, and Fixed Broadband subscriptions.


```python
trend_analysis = long_data.groupby(["Year", "Indicator"])["Value"].mean().reset_index()

trend_analysis
```

### 4.1 Trends and Patterns

The analysis shows a clear overall improvement in digital connectivity across the 54 African countries between 2010 and 2024.

- **Internet Users:** Average internet usage increased from 9.05% in 2010 to 44.89% in 2024, showing substantial growth in internet adoption.
- **Mobile Subscriptions:** Average mobile cellular subscriptions increased from 54.74 to 107.25 subscriptions per 100 people. This remained the strongest connectivity channel throughout the period.
- **Fixed Broadband:** Average fixed broadband subscriptions increased from 0.72 to 4.33 per 100 people. Although broadband adoption improved, it remained considerably lower than mobile connectivity.

Overall, the trend indicates that digital connectivity expanded significantly, with mobile networks playing a particularly important role in connecting people across Africa.

```python
relationship_data = long_data.pivot_table(
    index=["Country Name", "Year"],
    columns="Indicator",
    values="Value"
).reset_index()

relationship_data.head(10)
```

```python
correlation = relationship_data[
    ["Fixed Broadband", "Internet Users (%)", "Mobile Subscriptions"]
].corr()

correlation
```

### 4.2 Relationships between Variables

Correlation analysis was used to examine the relationships between the three digital connectivity indicators.

- **Internet Users and Mobile Subscriptions:** The correlation coefficient is **0.688**, indicating a moderate-to-strong positive relationship.
- **Internet Users and Fixed Broadband:** The correlation coefficient is **0.623**, indicating a moderate-to-strong positive relationship.
- **Fixed Broadband and Mobile Subscriptions:** The correlation coefficient is **0.471**, indicating a moderate positive relationship.

Overall, the results show that the digital connectivity indicators are positively related. The strongest relationship is between Internet Users and Mobile Subscriptions, suggesting that greater mobile connectivity is associated with higher internet adoption.

```python
change_over_time = relationship_data.groupby("Year")[
    ["Fixed Broadband", "Internet Users (%)", "Mobile Subscriptions"]
].mean()

change_over_time["Internet Users Change"] = (
    change_over_time["Internet Users (%)"].diff()
)

change_over_time["Fixed Broadband Change"] = (
    change_over_time["Fixed Broadband"].diff()
)

change_over_time["Mobile Subscriptions Change"] = (
    change_over_time["Mobile Subscriptions"].diff()
)

change_over_time
```

### 4.3 Significant Changes Over Time

The year-to-year analysis reveals several significant changes in digital connectivity between 2010 and 2024.

- **Internet Users:** The largest annual increase occurred in 2020, with an increase of **4.68 percentage points**. This indicates a notable acceleration in internet adoption during this period.
- **Fixed Broadband:** The largest annual increase occurred in 2024, rising by **0.94 subscriptions per 100 people**.
- **Mobile Subscriptions:** The largest annual increase occurred in 2024, increasing by **8.25 subscriptions per 100 people**.
- **Decline in Mobile Subscriptions:** A decrease of **1.59 subscriptions per 100 people** was observed in 2016.

Overall, digital connectivity generally followed an upward trend, although some indicators experienced temporary fluctuations. The strongest recent improvement occurred in 2024, particularly in mobile subscriptions and fixed broadband.

### 4.5 Meaningful Comparisons and Insights

A country-level comparison was conducted to identify differences in digital connectivity across the 54 African countries.

The analysis shows that **Morocco** has the highest average Internet Users rate at **69.11%**, followed by **Seychelles (64.80%)** and **South Africa (58.05%)**.

For Fixed Broadband, **Seychelles** has the highest average level at **18.98 subscriptions per 100 people**, followed closely by **Mauritius (18.80)** and **Tunisia (8.45)**.

An important finding is that higher internet usage does not necessarily correspond to higher fixed broadband adoption. For example, Morocco has the highest Internet Users rate but relatively lower fixed broadband adoption compared with Seychelles and Mauritius.

This suggests that internet access in many African countries may rely heavily on **mobile connectivity rather than fixed broadband infrastructure**.

Overall, the comparison highlights significant differences in digital connectivity between countries and suggests that expanding broadband infrastructure could help improve digital inclusion in countries where internet adoption is still relatively low.
